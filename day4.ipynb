```python
import numpy as np
```

```python
arr = np.arange(1, 31)
```

```python
matrix = arr.reshape(5, 6)
```

```python
print("Original Matrix:")
print(matrix)
```

```python
#extract third row
third_row = matrix[2]
print("Third Row:")
print(third_row)
```

```python
#extract third column
third_column = matrix[:, 2]
print("Third Column:")
print(third_column)
```

```python
#find all elements divisible by 3
divisible_by_3 = matrix[matrix % 3 == 0]
print("Elements divisible by 3:")
print(divisible_by_3)
```

```python
#replace all elements greater than 20 with 0
matrix[matrix > 20] = 0
print("Matrix after replacing elements greater than 20 with 0:")
print(matrix)
```

```python
#2. Create a Pandas DataFrame containing the names and marks of 8 students in three subjects: Python, DBMS, and Mathematics.

import pandas as pd
data = {
    "Name": ["Koustav", "Vansh", "Rugved", "Aditya", "Karan", "Shubham", "Piyush", "Rishi"],
    "Python": [75, 90, 78, 92, 88, 95, 80, 87],
    "DBMS": [92, 88, 85, 90, 86, 92, 84, 89],
    "Mathematics": [58, 92, 80, 95, 87, 98, 83, 90]
}

df = pd.DataFrame(data)
print(df)
```

```python
#calculate total marks of each student
df["Total Marks"] = df["Python"] + df["DBMS"] + df["Mathematics"]
print(df)
```

```python
#calculate the percentage
df["Percentage"] = (df["Total Marks"] / 300) * 100
print(df)
```

```python
#find the student with the highest percentage
highest_percentage_student = df.loc[df["Percentage"].idxmax()]
print("Student with the highest percentage:")
print(highest_percentage_student)
```

```python
# Display students having a percentage greater than 75.
high_performers = df[df["Percentage"] > 75]
print("Students with a percentage greater than 75:")
print(high_performers)
```

```python
#Sort the DataFrame according to percentage in descending order.
df_sorted = df.sort_values(by="Percentage", ascending=True)
print("DataFrame sorted by percentage in descending order:")
print(df_sorted)
```
