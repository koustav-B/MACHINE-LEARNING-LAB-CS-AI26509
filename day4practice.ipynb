```python
import pandas as pd
df = pd.read_csv("sales.csv")
print(df)
```

```python
# (b) Display the first five records
print("First five records:")
print(df.head())
```

```python
# (c) Check for missing values
print("\nMissing values:")
print(df.isnull().sum())
```

```python

df["Sales"] = df["Quantity"] * df["Price"]
print("\nDataFrame with Sales:")
print(df)
```

```python
# (e) Find total sales
total_sales = df["Sales"].sum()

print("\nTotal Sales:", total_sales)
```

```python

highest_sales = df.loc[df["Sales"].idxmax()]
print("\nProduct with highest sales:")
print(highest_sales["Products"])
print("Sales:", highest_sales["Sales"])
```

```python
# (g) Find total sales for each category
category_sales = df.groupby("Category")["Sales"].sum()

print("\nTotal sales for each category:")
print(category_sales)
```

```python
# (h) Sort records by Sales in descending order
sorted_df = df.sort_values(by="Sales", ascending=False)

print("\nRecords sorted by Sales in descending order:")
print(sorted_df)
```

```python
import seaborn as sns
import pandas as pd

# Load the Tips dataset
tips = sns.load_dataset("tips")

# (a) Display first 50 observations
print("First 50 observations:")
print(tips.head(50))
```

```python
# (b) Check the type of data structure
print("\nType of data structure:")
print(type(tips))
```

```python
# (c) Check the types of features
print("\nData types of features:")
print(tips.dtypes)
```

```python
# (d) Describe the dataset
print("\nDescription of the dataset:")
print(tips.describe())
```

```python
# Five-number summary
print("\nFive-number summary:")
print(tips.describe().loc[["min", "25%", "50%", "75%", "max"]])
```

```python
# Count, Mean and Standard Deviation
print("\nCount, Mean and Standard Deviation:")
print(tips.describe().loc[["count", "mean", "std"]])
```

```python
# (e) Display detailed observations during Sunday
print("\nObservations during Sunday:")
print(tips[tips["day"] == "Sun"])
```

```python
# (f) Create sun_tips DataFrame
sun_tips = tips[tips["day"] == "Sun"].copy()

print("\nSunday Tips DataFrame:")
print(sun_tips)
```

```python
# (g) Create bill_per_person feature
sun_tips["bill_per_person"] = sun_tips["total_bill"] / sun_tips["size"]

print("\nUpdated Sunday DataFrame:")
print(sun_tips)
```

```python
# (h) Calculate tips percentage by every customer on Sunday   
sun_tips["tip_percentage"] = (sun_tips["tip"] / sun_tips["total_bill"]) * 100

print("\nTip percentage for every customer on Sunday:")
print(sun_tips[["total_bill", "tip", "tip_percentage"]])
```

```python
# (i) Compute average percentage tips from the dataset
tips["tip_percentage"] = (tips["tip"] / tips["total_bill"]) * 100

average_percentage = tips["tip_percentage"].mean()

print("\nAverage percentage tip from the dataset:")
print(average_percentage)
```
