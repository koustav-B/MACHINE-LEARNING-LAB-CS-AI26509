```python
# 1. Create a list 'number' containing 10, 20, 30, 40, 50
number = [10, 20, 30, 40, 50]
print("1. Original list:", number)
```

```python
number.append(60)
number.insert(2, 25)
print("2. List after insertion:", number)
```

```python
number.remove(30)
print("3. Final list after removing 30:", number)
```

```python
# 4. Find the largest and smallest number
largest = max(number)
smallest = min(number)

print("4. Largest number:", largest)
print("   Smallest number:", smallest)
```

```python
# 5. Reverse the list and find its sum
reversed_number = number[::-1]
sum_reversed = sum(reversed_number)

print("5. Reversed list:", reversed_number)
print("   Sum of reversed list:", sum_reversed)
```

```python
# 6. Create a list of squares from 10 to 20 using list comprehension
squares = [x**2 for x in range(10, 21)]
print("6. Squares from 10 to 20:", squares)
```

```python
# 14. Create a DataFrame with Name and Marks
import pandas as pd

data = {
    "Name": ["Amit", "Rahul", "Priya", "Sneha", "Karan"],
    "Marks": [85, 72, 95, 65, 40]
}

df = pd.DataFrame(data)

print("14. Data Frame:")
print(df)
```

```python

# 15. Display first two rows and only Marks column
print("15. First two rows:")
print(df.head(2))

print("Only Marks column:")
print(df["Marks"])
```

```python
# 16. Add a Grade column
def calculate_grade(marks):
    if marks >= 90:
        return "A+"
    elif marks >= 75:
        return "A"
    elif marks >= 50:
        return "B"
    else:
        return "C"

df["Grade"] = df["Marks"].apply(calculate_grade)

print("16. Data Frame after adding Grade:")
print(df)
```

```python
# 17. Find average marks and students scoring more than 35
average_marks = df["Marks"].mean()

students_above_35 = df[df["Marks"] > 35]

print("17. Average marks:", average_marks)

print("Students scoring more than 35:")
print(students_above_35)
```

```python
# 18. Save the DataFrame into a CSV file
df.to_csv("student_marks.csv", index=False)

print("18. DataFrame saved successfully as 'student_marks.csv'")
```
